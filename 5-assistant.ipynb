```python
import os
from google import genai
from google.genai import types
from dotenv import load_dotenv

# Carrega as variáveis de ambiente
load_dotenv()
```

```python
# Inicializa o cliente do Gemini
client = genai.Client(api_key=os.environ.get("GEMINI_API_KEY"))
MODEL_GEMMA4 = "gemma-4-26b-a4b-it"
```

```python
# No Gemini, as instruções e ferramentas são configuradas na sessão de chat.
instructions = "Você é um tutor sobre assuntos relacionados a tecnologia"
tools = [types.Tool(code_execution=types.ToolCodeExecution())]

chat = client.chats.create(
    model=MODEL_GEMMA4,
    config=types.GenerateContentConfig(
        system_instruction=instructions,
        tools=tools,
        temperature=0.1
    )
)
```

```python
pergunta = """
 Em um sistema de cache com mapeamento direto, dado uma memória principal de uma cache
 de 256 KB e blocos de 64 bytes:
 - Quantas linhas existem na cache?
 - Onde o bloco de endereço 0x1A23C será mapeado na cache?
"""
```

```python
# Envia a mensagem e aguarda o processamento (o Gemini executará o código se necessário)
response = chat.send_message(pergunta)
```

```python
# Exibe a resposta final do modelo
print(response.text)
```

```python
# Analisando os passos da execução de código (equivalente aos run_steps da OpenAI)
for part in response.candidates[0].content.parts:
    if part.executable_code:
        print("\n==== Código Executado ====")
        print(part.executable_code.code)
    if part.code_execution_result:
        print("\n==== Resultado da Execução ====")
        print(part.code_execution_result.output)
    if part.text and not (part.executable_code or part.code_execution_result):
        print("\n==== Resposta de Texto ====")
        print(part.text)
```
