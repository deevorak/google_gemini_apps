```python
from google import genai
from google.genai import types
import os
from PIL import Image
import requests
import io
```

```python
client = genai.Client(api_key=os.environ.get("GEMINI_API_KEY"))
```

### Gerar Imagens

```python
nome="robo_praia"
model="gemini-2.5-flash"
prompt="Crie uma imagem com dois robôs em uma praia lendo livro de finanças"
```

```python
resposta = client.models.generate_images(
    model=model,
    prompt=prompt,
    config=types.GenerateImagesConfig(
        number_of_images=1
    )
)
```

```python
# O Imagen não retorna um prompt revisado da mesma forma que o DALL-E.
```

```python
nome_arquivo = f"files/{nome}_{model}.jpg"
image_bytes = resposta.generated_images[0].image.image_bytes
with open(nome_arquivo, "wb") as f:
    f.write(image_bytes)

from IPython.display import Image as IPImage
IPImage(filename=nome_arquivo)
```

### Image to Text

```python
url = "https://harus.ind.br/wp-content/uploads/2023/02/5-praias-brasileiras.jpg"
image_data = requests.get(url).content
image = Image.open(io.BytesIO(image_data))

resposta = client.models.generate_content(
    model="gemini-2.5-flash",
    contents=["Descreva a imagem fornecida", image]
)
```

```python
print(resposta.text)
```

### Extraia placa de carro

```python
texto = "Qual é a placa? Devolva apenas a placa!"
url = "https://quatrorodas.abril.com.br/wp-content/uploads/2018/09/imagem-e1536873662306.jpg?quality=70&strip=info&w=720&crop=1"
image_data = requests.get(url).content
image = Image.open(io.BytesIO(image_data))
```

```python
resposta = client.models.generate_content(
    model="gemini-2.5-flash",
    contents=[texto, image]
)
```

```python
print(resposta.text)
```

```python
url = "https://upload.wikimedia.org/wikipedia/commons/thumb/d/dd/Gfp-wisconsin-madison-the-nature-boardwalk.jpg/2560px-Gfp-wisconsin-madison-the-nature-boardwalk.jpg"
image_data = requests.get(url).content
image = Image.open(io.BytesIO(image_data))

response = client.models.generate_content(
    model="gemini-1.5-flash",
    contents=[
        "What are in these images? Is there any difference between them?",
        image,
        image 
    ]
)
```
