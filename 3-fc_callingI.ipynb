```python
import json
import os
from google import genai
from google.genai import types
from dotenv import load_dotenv

load_dotenv()
```

```python
client = genai.Client(api_key=os.environ.get("GEMINI_API_KEY"))
MODEL_GEMMA4 = "gemma-4-26b-a4b-it"
```

```python
def saudacao_por_periodo(hora: int) -> str:
    """Retorna uma saudação baseada na hora do dia (formato 24h)."""
    if 5 <= hora < 12:
        return json.dumps({"saudacao": "Bom dia!"})
    elif 12 <= hora < 18:
        return json.dumps({"saudacao": "Boa tarde!"})
    elif 18 <= hora < 22:
        return json.dumps({"saudacao": "Boa noite!"})
    else:
        return json.dumps({"saudacao": "Boa madrugada!"})
```

```python
# No SDK do Gemini, passamos a função diretamente como ferramenta
tools = [saudacao_por_periodo]

# Inicializa o chat com a configuração de ferramentas
chat = client.chats.create(
    model=MODEL_GEMMA4,
    config=types.GenerateContentConfig(
        tools=tools,
        temperature=0
    )
)
```

```python
mensagem = "Qual saudação o modelo me dá se for 14h?"
print(f"Usuário: {mensagem}")

response = chat.send_message(mensagem)
print(f"\nGemini: {response.text}")
```
