```python
import os
import time
from google import genai
from google.genai import types
from dotenv import load_dotenv

load_dotenv()
```

```python
client = genai.Client(api_key=os.environ.get("GEMINI_API_KEY"))
MODEL_GEMMA4 = "gemma-4-26b-a4b-it"
```

```python
# Upload do arquivo
# No Gemini, não precisamos criar um Vector Store explicitamente para um único arquivo
file_path = "files/LLM.pdf"
file = client.files.upload(file=file_path)
print(f"Arquivo enviado: {file.display_name}")
```

```python
# Aguarda o processamento do arquivo
while file.state.name == "PROCESSING":
    print(".", end="")
    time.sleep(2)
    file = client.files.get(name=file.name)

print(f"\nStatus do arquivo: {file.state.name}")
```

```python
# Configuração do Assistente (Tutor)
instructions = """Você é um tutor especializado em tecnologias emergentes. 
Você sabe responder perguntas sobre LLMs como OpenAI, HuggingFace, etc. 
Caso você não encontre as respostas no documento, seja sincero e fale que não sabe responder"""

chat = client.chats.create(
    model=MODEL_GEMMA4,
    config=types.GenerateContentConfig(
        system_instruction=instructions,
        temperature=0.1
    )
)
```

```python
pergunta = "Conforme o documento, o que é o OpenAI?"
```

```python
# Envia a pergunta junto com a referência ao arquivo
response = chat.send_message([pergunta, file])
```

```python
# Verifica a resposta
if response.text:
    print(response.text)
else:
    print("Não foi possível obter uma resposta.")
```
