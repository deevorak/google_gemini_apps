## Fundos Imobiliários

```python
from dotenv import load_dotenv
from google import genai
from google.genai import types
import os
load_dotenv()

client = genai.Client(api_key=os.environ.get("GEMINI_API_KEY"))
```

```python
# MODEL_ID_LEGACY_FASTEST="gemini-2.5-flash-lite" # mais barato
# MODEL_ID_LEGACY_FAST="gemini-2.5-flash"
# MODEL_ID_LEGACY_DEEP="gemini-2.5-pro"
# MODEL_ID_FASTEST= "gemini-3.1-flash-lite-preview"
# MODEL_ID_FAST="gemini-3.1-flash-preview"
# MODEL_ID_DEEP="gemini-3.1-pro-preview"
MODEL_GEMMA4="gemma-4-26b-a4b-it"

system_instruction = """Você é um analista financeiro especializado em fundos imobiliários.
Faça sempre uma analise com base nos últimos 5 fatos relevantes publicados pela gestora de um fundo imobiliário.
Para recomendações use os fatos relevantes apurados e indicadores correntes mais comuns.
Use a internet para pesquisar valores atuais em sites especializados em mercado de ações B3.
"""

prompt = """
<request>
Faça uma análise do fundo imobiliário RBRY11 usando os indicadores disponíveis em https://investidor10.com.br/fiis/rbry11 e os fatos relevantes disponíveis em https://realestate.patria.com/papel/rbry11/documentos para o ano de 2026." \
Faça recomendação de compra, venda ou manutenção.
</request>
<format>
    COTAÇÃO: [valor cotação do dia]
    VP: [valor patrimonial da cota]
    P/VP: [preço / valor patrimonial]
    DY 12 meses: [Dividend yeld nos últimos 12 meses]
    
    RESUMO: [resumo da análise em até 5 linhas sobre como está a situação do fundo com base nos fatos relevantes]

    PONTOS DE ATENÇÃO: [Listar pontos de atenção]
    RECOMENDAÇÃO: [COMPRA/NEUTRO/VENDA]
</format>
"""

grounding_tool = types.Tool(
    google_search=types.GoogleSearch()
)

response = client.models.generate_content(
    model=MODEL_GEMMA4,
    contents=prompt,
    config=types.GenerateContentConfig(
        temperature=1,
        system_instruction=system_instruction,
        tools=[grounding_tool]
    )
)

print(response.text)
```
